```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])
```

```python
df.info()
```

```python
df.head()
```

```python
df["Length of Stay"] = (
    df["Discharge Date"] - df["Date of Admission"]
).dt.days
```

```python
df["Admission Year"] = df["Date of Admission"].dt.year
```

```python
df["Admission Month"] = df["Date of Admission"].dt.month_name()
```

```python
df.head()
```

```python
df["Age Group"] = pd.cut(
    df["Age"],
    bins=[0, 18, 30, 45, 60, 100],
    labels=["0-18", "19-30", "31-45", "46-60", "60+"]
)
```

```python
df["Billing Category"] = pd.qcut(
    df["Billing Amount"],
    q=3,
    labels=["Low", "Medium", "High"]
)
```

```python
monthly_admissions = (
    df.groupby("Admission Month")
      .size()
      .reindex([
          "January","February","March","April","May","June",
          "July","August","September","October","November","December"
      ])
)

monthly_admissions
```

```python
monthly_admissions.plot(kind="bar", figsize=(10,5))
plt.title("Monthly Admissions")
plt.xlabel("Month")
plt.ylabel("Patients")
plt.xticks(rotation=45)
plt.show()
```

```python
condition_revenue = (
    df.groupby("Medical Condition")["Billing Amount"]
      .sum()
      .sort_values(ascending=False)
)

condition_revenue
```

```python
condition_revenue.plot(kind="bar", figsize=(8,5))
plt.title("Revenue by Medical Condition")
plt.ylabel("Billing Amount")
plt.xticks(rotation=45)
plt.show()
```
